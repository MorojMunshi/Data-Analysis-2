## Q1 — Basic Processing & Word Frequencies (Lecture 1) — 1 point
**Task:**  
1) Read the corpus into DataFrame `df` if not already loaded.  
2) Create `text_norm` (lowercase, remove diacritics/tatweel/punctuation, unify alef/yeh).  
3) Tokenize `text_norm` into words `tokens`.  
4) Compute **top 10 Arabic tokens** (length > 1) over Arabic lines only → `top10_ar`.  



````python
import re

def normalize_arabic(text):
    # إزالة التشكيل
    text = re.sub(r"[\u064B-\u0652]", "", text)
    # إزالة التطويل
    text = re.sub(r"ـ", "", text)
    # توحيد الألف
    text = re.sub(r"[إأآا]", "ا", text)
    # توحيد الياء
    text = re.sub(r"[يى]", "ي", text)
    # إزالة الترقيم
    text = re.sub(r"[^\w\s]", "", text)
    # lowercase
    return text.lower().strip()

df["text_norm"] = df["text"].apply(normalize_arabic)
df[["text", "text_norm"]].head()
````

````python
# 3
import re

def tokenize(text):
    # استخراج الكلمات العربية والإنجليزية فقط
    return re.findall(r"[\u0621-\u063A\u0641-\u064A]+|[a-zA-Z]+", text)

# إنشاء عمود جديد tokens
df["tokens"] = df["text_norm"].apply(tokenize)

# عرض النتيجة في جدول clean + tokens
df[["text_norm","tokens"]].head().rename(columns={"text_norm":"clean"})
````

````python
#4
from sklearn.feature_extraction.text import CountVectorizer
import pandas as pd

# نحول العمود tokens إلى نصوص مفصولة بمسافات (للأسطر العربية فقط)
def to_docs(tokens_col):
    subset = df[df['lang']=='ar']            # فقط النصوص العربية
    return [' '.join(toks) for toks in subset[tokens_col]]

# CountVectorizer لاستخراج unigram عربي فقط (طوله > 1)
cv_uni = CountVectorizer(
    ngram_range=(1,1),
    token_pattern=r'[\u0621-\u063A\u0641-\u064A]{2,}'  # أحرف عربية فقط، طول ≥ 2
)

X_uni = cv_uni.fit_transform(to_docs('tokens'))

# حساب التكرارات
uni_counts = pd.Series(
    X_uni.toarray().sum(axis=0),
    index=cv_uni.get_feature_names_out()
).sort_values(ascending=False)

# أعلى 10 كلمات عربية
top10_ar = uni_counts.head(10).reset_index()
top10_ar.columns = ["token","count"]

print(top10_ar)
````

## Q2 — Stemming vs. Lemmatization  — 1 point
**Task:**  
1) Create two functions:  
   - `stem_list_ar(tokens)` → stems (e.g., NLTK ISRIStemmer).  
   - `lemma_list_ar(tokens)` → lemmas (approximate if no Arabic lemmatizer available).  
2) Apply both to tokens of row `id=2` → `stems_id2`, `lemmas_id2`.  
3) Show (token, stem, lemma) for up to 8 items.


````python
#1
from nltk.stem.isri import ISRIStemmer
import pandas as pd
import re

# محلل صرفي عربي
stemmer_ar = ISRIStemmer()

# دالة استخراج الجذور
def stem_list_ar(tokens):
    return [stemmer_ar.stem(t) for t in tokens]

# دالة Lemmatization تقريبية (ترجع نفس الكلمة)
def lemma_list_ar(tokens):
    return tokens

# تطبيق الدوال على العمود tokens
df["stems_ar"]  = df["tokens"].apply(stem_list_ar)
df["lemmas_ar"] = df["tokens"].apply(lemma_list_ar)

# تفجير القوائم بحيث كل كلمة + جذرها + lemma في صف مستقل
df_exploded = df.explode(["tokens","stems_ar","lemmas_ar"])

# فلترة: إبقاء الكلمات العربية فقط
arabic_only = df_exploded[df_exploded["tokens"].str.match(r'^[\u0621-\u063A\u0641-\u064A]+$')]

# عرض الجدول النهائي
display(arabic_only[["tokens","stems_ar","lemmas_ar"]])
````

````python
# استخراج التوكنز للصف id=2
tokens_id2 = df.loc[df["id"]==2, "tokens"].values[0]

# تطبيق Stemming و Lemmatization
stems_id2  = stem_list_ar(tokens_id2)
lemmas_id2 = lemma_list_ar(tokens_id2)

# عرض النتيجة في جدول مرتب
result_id2 = pd.DataFrame({
    "tokens": tokens_id2,
    "stems_id2": stems_id2,
    "lemmas_id2": lemmas_id2
})

# لو في Jupyter/Colab
display(result_id2)
````

````python
# نطبق دوال stemming و lemmatization على العمود tokens
df["stems_ar"]  = df["tokens"].apply(stem_list_ar)
df["lemmas_ar"] = df["tokens"].apply(lemma_list_ar)

# نفرد القوائم كلمة كلمة
df_exploded = df.explode(["tokens","stems_ar","lemmas_ar"])

# نعرض فقط أول 8 عناصر كجدول مرتب
result_8 = df_exploded[["tokens","stems_ar","lemmas_ar"]].head(8)
result_8 = result_8.rename(columns={"tokens":"token","stems_ar":"stem","lemmas_ar":"lemma"})

display(result_8)
````

## Q3 — TF‑IDF Feature Exploration  — 1 point
**Task:**  
1) Fit `TfidfVectorizer(ngram_range=(1,2))` on `df["text_norm"]`.  
2) For document `id=1`, extract top **10** features by TF‑IDF weight → `top10_tfidf_id1` as a list of `(feature, weight)`.

````python
from sklearn.feature_extraction.text import TfidfVectorizer

# ندرّب TF-IDF على النصوص المنظّفة
tfidf = TfidfVectorizer(ngram_range=(1,2))
X = tfidf.fit_transform(df["text_norm"])
feature_names = tfidf.get_feature_names_out()

# دالة لاستخراج أعلى k ميزات
def top_terms(row_idx, k=10):
    row = X[row_idx].toarray().ravel()
    topk = row.argsort()[-k:][::-1]
    return [(feature_names[i], float(row[i])) for i in topk]

# نجيب index للوثيقة id=1
row_idx = df.index[df["id"]==1][0]

# نطلع Top 10 فقط لها
top10_tfidf_id1 = top_terms(row_idx, 10)

print("Document id=1:", df.loc[row_idx, "text"])
print("Top 10 TF-IDF features:\n", top10_tfidf_id1)
````

##  Q4 — POS + Dependency Parse  — 1 point
**Task:**  
1) Using `stanza` (Arabic), POS-tag and dependency-parse: `"الثعلب السريع قفز فوق البيت"`.  
2) Save POS tags to `pos_ar` as `(token, upos)` tuples.  
3) Save dependencies to `deps_ar` as `(head, dep, relation)` tuples.

````python
!pip install stanza
````

````python

# Your Answer — Q4

import stanza
import pandas as pd

# تحميل موديل اللغة العربية (مرة واحدة فقط)
stanza.download("ar")
nlp = stanza.Pipeline("ar", processors="tokenize,pos,lemma,depparse")

# الجملة
text = "الثعلب السريع قفز فوق البيت"
doc = nlp(text)

# استخراج POS tags
pos_ar = [(w.text, w.upos) for sent in doc.sentences for w in sent.words]

# استخراج dependency relations
deps_ar = [
    (sent.words[w.head-1].text if w.head > 0 else "ROOT", w.text, w.deprel)
    for sent in doc.sentences for w in sent.words
]

# تحويل إلى جداول
df_pos = pd.DataFrame(pos_ar, columns=["Token", "POS"])
df_deps = pd.DataFrame(deps_ar, columns=["Head", "Dep", "Relation"])

print("POS Tags Table:")
print(df_pos, "\n")

print("Dependencies Table:")
print(df_deps)
````

## Q5 — Quick Sentiment (VADER— 1 point
**Task:**  
1) Use **VADER** from NLTK to compute sentiment scores for the **English** lines only.  
2) Store a list `vader_scores` of `(id, compound)` tuples sorted by id.


````python
from nltk.sentiment.vader import SentimentIntensityAnalyzer

# إنشاء محلل VADER
sia = SentimentIntensityAnalyzer()

vader_scores = []
for row in df[df["lang"]=="en"].itertuples():
    scores = sia.polarity_scores(row.text)
    compound = scores["compound"]

    # التصنيف حسب القيمة
    if compound >= 0.05:
        label = "positive"
    elif compound <= -0.05:
        label = "negative"
    else:
        label = "neutral"

    vader_scores.append((row.id, compound, label))

# ترتيب حسب id
vader_scores = sorted(vader_scores, key=lambda x: x[0])

# عرض النتيجة
for vid, comp, lab in vader_scores:
    print(f"id={vid} → {lab} (compound={comp:.3f})")
````
